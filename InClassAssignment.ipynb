```python
def getValues(n):
    values = []
    for i in range(n):
        try:
            value = int(input(f"Enter value {i+1}: "))
        except ValueError:
            print("Invalid input. Please enter a valid figure.")
        values.append(value)
    return values


def calcAverage(aList):
    try:
        if len(aList) <= 2:
            print("Division by zero is not allowed.")
            return None
        else:
            relevant_values = aList[2:]
            average = sum(relevant_values) / len(relevant_values)
            return average
    except:
        return None


if __name__ == "__main__":
    try:
        n = int(input("Enter the total number of values: "))
        aList = getValues(n)
        average = calcAverage(aList)
        print(average)
    except ValueError:
        print("Invalid input. Please enter a valid input.")
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, female=False):
        if female:
            note = f"Ms. {self.first} {self.last}"
        else:
            note = f"Mr. {self.first}"
        print(note)

    def __str__(self):
        return str(self.first) + " " + str(self.last)
        return "{} {}".format(self.first, self.last)
        return f"{self.first} {self.last}"

oneGreeting = Greeting("Maria", "Becker")
oneGreeting.display(female=True)
print(oneGreeting)
```

```python
class Student:
    def __init__(self, name, number):
        self.name = name
        self.number = number
        self.courses = []

    def enroll(self, course):
        if course not in self.courses:
            self.courses.append(course)
        else:
            return f"{self.name} is already enrolled in {course}."

    def get_courses(self):
        return self.courses


s1 = Student("James Bond", 1001)
s2 = Student("John Wick", 1002)

s1.name, s2.name
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)

corner_1 = Point(0, 0)
corner_2 = Point(2, 0)
corner_3 = Point(0, 2)
corner_4 = Point(2, 2)

corner_1.distance(corner_2)
```

```python
class Rectangle:
    def __init__(self):
        self.c1 = Point(0, 0)
        self.c2 = Point(2, 0)
        self.c3 = Point(0, 2)
        self.c4 = Point(2, 2)
        self.width = self.calculate_width()
        self.length = self.calculate_length()

    def calculate_width(self):
        return self.c1.distance(self.c2)

    def calculate_length(self):
        return self.c1.distance(self.c3)

    def area(self):
        return f"Area: {self.width * self.length} sq. ft"

square = Rectangle()
square.area()
```
